## 🧪 Interactive Demo 1: How NEXUS Calculates Orders & GST

Let's see how NEXUS handles the math for a commercial order with:
- Multiple products
- **Dual-Mode Discounts** (e.g., Flat ₹500 off vs. 10% off)
- **GST Tax Computation** (18% GST split into CGST + SGST)

````python
# Python demonstration of NEXUS pricing & GST engine
def calculate_nexus_order(items, discount_val, discount_type='fixed', gst_rate=18.0):
    subtotal = sum(item['qty'] * item['price'] for item in items)
    
    # Dual-mode discount logic
    if discount_type == 'percent':
        discount_amount = round(subtotal * (discount_val / 100.0), 2)
    else:
        discount_amount = min(subtotal, float(discount_val))
        
    taxable_amount = max(0.0, subtotal - discount_amount)
    gst_amount = round(taxable_amount * (gst_rate / 100.0), 2)
    grand_total = round(taxable_amount + gst_amount, 2)
    
    # Tax split
    cgst = round(gst_amount / 2.0, 2)
    sgst = round(gst_amount / 2.0, 2)
    
    print('========================================')
    print('        NEXUS ORDER SUMMARY')
    print('========================================')
    for item in items:
        line_total = item['qty'] * item['price']
        print(f"• {item['name']:<24} {item['qty']}x @ ₹{item['price']:>7,.2f} = ₹{line_total:>9,.2f}")
    print('----------------------------------------')
    print(f"Subtotal:                     ₹{subtotal:>10,.2f}")
    print(f"Discount ({discount_type}):              -₹{discount_amount:>10,.2f}")
    print(f"Taxable Amount:               ₹{taxable_amount:>10,.2f}")
    print(f"CGST ({gst_rate/2}%):                  +₹{cgst:>10,.2f}")
    print(f"SGST ({gst_rate/2}%):                  +₹{sgst:>10,.2f}")
    print('========================================')
    print(f"GRAND TOTAL PAYABLE:          ₹{grand_total:>10,.2f}")
    print('========================================\n')

# Sample Customer Order
sample_cart = [
    {'name': 'Alphonso Mango Crate', 'qty': 2, 'price': 3200.0},
    {'name': 'Filter Coffee Concentrate', 'qty': 1, 'price': 1800.0},
    {'name': 'Kaju Badam Dry Fruits', 'qty': 2, 'price': 650.0}
]

# Calculate with ₹500 Flat Discount & 18% GST
calculate_nexus_order(sample_cart, discount_val=500, discount_type='fixed', gst_rate=18.0)
````

## 📦 Interactive Demo 2: Zero-Overselling Warehouse Stock Lock

In traditional apps, two customers buying the last item at the same time causes negative stock.
**NEXUS uses Atomic Row Locking (`SELECT ... FOR UPDATE`)** in PostgreSQL to guarantee stock safety.

````python
class NexusWarehouse:
    def __init__(self):
        # Live warehouse stock database simulation
        self.inventory = {
            ('NX-MNG-001', 'Mumbai'): 10,
            ('NX-MNG-001', 'Delhi'): 5,
            ('NX-CFE-001', 'Mumbai'): 25
        }
    
    def process_order(self, sku, warehouse, requested_qty, customer_name):
        key = (sku, warehouse)
        available = self.inventory.get(key, 0)
        
        print(f"[TX START] Customer '{customer_name}' attempting to buy {requested_qty}x of {sku} from {warehouse}...")
        
        # Atomic Concurrency Check (Simulating PostgreSQL Row Lock)
        if available < requested_qty:
            print(f"❌ REJECTED (HTTP 400): Insufficient stock in {warehouse}. Available: {available}, Requested: {requested_qty}.")
            print("   [TRANSACTION ROLLED BACK - Stock untouched]\n")
            return False
        
        # Decrement stock safely
        self.inventory[key] -= requested_qty
        print(f"✅ APPROVED (HTTP 201): Order processed! Stock in {warehouse} updated: {available} -> {self.inventory[key]}.")
        print("   [TRANSACTION COMMITTED - Ledger entry recorded]\n")
        return True

wh = NexusWarehouse()

# Customer 1 orders 8 crates (Successful)
wh.process_order('NX-MNG-001', 'Mumbai', 8, 'Riya Mehta')

# Customer 2 orders 5 crates (Rejected because only 2 left!)
wh.process_order('NX-MNG-001', 'Mumbai', 5, 'Priya Nair')
````

## 🔐 Interactive Demo 3: Cryptographic QR Invoices (Tamper-Proof)

Every invoice issued by NEXUS has a **Digital HMAC-SHA256 Signature**.
If anyone changes even 1 single rupee on the bill, the digital signature breaks immediately!

````python
import hmac
import hashlib

# Secret key known only to the NEXUS secure server
NEXUS_SERVER_SECRET = b'nexus-super-secure-production-secret-key-2026'

def generate_invoice_signature(invoice_number):
    data = f"invoice:{invoice_number}".encode('utf-8')
    sig = hmac.new(NEXUS_SERVER_SECRET, data, hashlib.sha256).hexdigest()[:32]
    return sig

def verify_invoice_qr(invoice_number, scanned_signature):
    expected = generate_invoice_signature(invoice_number)
    # Constant-time comparison to prevent timing attacks
    return hmac.compare_digest(expected, scanned_signature)

# 1. Server generates invoice and embeds QR code
inv_num = 'NEXUS-INV-2026-00042'
genuine_sig = generate_invoice_signature(inv_num)
qr_url = f"https://nexus.enterprise/api/invoices/public/{inv_num}/{genuine_sig}.pdf"

print(f"📄 Genuine Invoice:    {inv_num}")
print(f"🔑 Embedded HMAC Sig:  {genuine_sig}")
print(f"🔗 QR Verification URL: {qr_url}")
print(f"🔍 Verification Check: {verify_invoice_qr(inv_num, genuine_sig)} (VERIFIED ORIGINAL)\n")

# 2. Attacker tries to forge the invoice
fake_sig = 'abc123fakefakefakefakefake456789'
print(f"🚨 Fake Check Test:    {verify_invoice_qr(inv_num, fake_sig)} (REJECTED FORGERY!)")
````

## 🧠 Interactive Demo 4: AI Predictive Analytics & Forecasting

NEXUS runs an ordinary least squares (OLS) linear regression model over 14-day sales velocity to predict **forward 30-day cash flow** and alert when products will run out of stock.

````python
# Simple OLS linear regression forecast simulation
history_daily_revenue = [2800, 3100, 3050, 3300, 3450, 3200, 3600, 3750, 3900, 3850, 4100, 4300, 4250, 4500]

n = len(history_daily_revenue)
x = list(range(n))
y = history_daily_revenue

sum_x = sum(x)
sum_y = sum(y)
sum_xy = sum(i * j for i, j in zip(x, y))
sum_xx = sum(i * i for i in x)

slope = (n * sum_xy - sum_x * sum_y) / (n * sum_xx - sum_x ** 2)
intercept = (sum_y - slope * sum_x) / n

# Predict day 15
next_day_prediction = intercept + slope * n
projected_30_day_cash_flow = next_day_prediction * 30

print('==============================================')
print('      NEXUS FASTAPI AI FORECAST ENGINE')
print('==============================================')
print(f"14-Day Baseline Average: ₹{sum_y/n:>10,.2f} / day")
print(f"Growth Slope Velocity:   +₹{slope:>10,.2f} / day")
print(f"Projected Tomorrow:      ₹{next_day_prediction:>10,.2f}")
print(f"Forward 30-Day Forecast: ₹{projected_30_day_cash_flow:>10,.2f}")
print(f"Trend Indicator:         RISING (High Confidence)")
print('==============================================')

# Stock Run-Out Matrix
current_stock = 45
daily_demand = 8
days_remaining = current_stock / daily_demand
risk_level = 'CRITICAL' if days_remaining <= 7 else 'HIGH' if days_remaining <= 14 else 'OPTIMAL'

print(f"\nStock Intelligence: {current_stock} units left at {daily_demand}/day = {days_remaining:.1f} days remaining ({risk_level} ALERT)")
````

## 💰 Interactive Demo 5: True Net Profit vs. Gross Revenue

NEXUS tracks Cost of Goods Sold (COGS), operating expenses (rent, electricity, logistics), and employee salaries to calculate **true net profit**.

````python
# Financial Breakdown Simulation
gross_revenue = 482900.00         # Total sales collected
cost_of_goods_sold = 215000.00    # What the products cost to purchase
gross_profit = gross_revenue - cost_of_goods_sold

operating_expenses = 84000.00     # Warehouse rent, packaging, electricity
employee_salaries = 95000.00      # Staff payroll
total_spending = operating_expenses + employee_salaries

net_profit = gross_profit - total_spending
profit_margin = (net_profit / gross_revenue) * 100.0

print('==================================================')
print('        NEXUS COMPANY FINANCIAL INTELLIGENCE')
print('==================================================')
print(f"Gross Sales Revenue:          ₹{gross_revenue:>12,.2f}")
print(f"Cost of Goods Sold (COGS):   -₹{cost_of_goods_sold:>12,.2f}")
print('--------------------------------------------------')
print(f"GROSS PROFIT:                 ₹{gross_profit:>12,.2f}")
print(f"Operating Expenses:          -₹{operating_expenses:>12,.2f}")
print(f"Employee Salaries:           -₹{employee_salaries:>12,.2f}")
print('==================================================')
print(f"REAL NET PROFIT:              ₹{net_profit:>12,.2f} ({profit_margin:.1f}% Margin)")
print('==================================================')
````
